## 2. Industrial Surface Defect Catalog

The benchmark dataset encapsulates six real-world industrial defect types alongside defect-free standard reference sheets:
- **Crazing (`Cr`)**: Thermal microfissuring
- **Inclusion (`In`)**: Embedded foreign oxide slag particles
- **Patches (`Pa`)**: Localized surface plate delamination
- **Pitted Surface (`PS`)**: Chemical/mechanical oxidative etch pits
- **Rolled-in Scale (`RS`)**: Iron oxide entrapment during rolling
- **Scratches (`Sc`)**: Severe mechanical longitudinal guide gouges
- **Normal Reference**: Pristine surface finish
